# Vyom-JEPA Analysis

This notebook analyzes the training logs and model checkpoints.


```python
import pandas as pd
import matplotlib.pyplot as plt
import torch
import sys
import os

# Add src to path
sys.path.append(os.path.abspath("../src"))

from vyom_jepa.models.vyom_jepa import VyomJEPA
```

## 1. Training Curves


```python
run_name = "run_v1"  # Change this to match config.yaml
base_path = f"../experiments/{run_name}"

try:
    df = pd.read_csv(f"{base_path}/training_log.csv")
    plt.figure(figsize=(10, 5))
    plt.plot(df["Epoch"], df["Loss"], marker="o")
    plt.title(f"Training Loss - {run_name}")
    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.grid(True)
    plt.show()
except FileNotFoundError:
    print(f"No training logs found at {base_path}. Run training first!")
```

## 2. Load Model Checkpoint


```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = VyomJEPA().to(device)

checkpoint_path = f"{base_path}/checkpoints/model_epoch_5.pt"
if os.path.exists(checkpoint_path):
    model.load_state_dict(torch.load(checkpoint_path))
    print("Model loaded successfully!")
else:
    print(f"Checkpoint not found at {checkpoint_path}.")
```
